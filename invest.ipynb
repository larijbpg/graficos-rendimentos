```python

# Objetivo: Criar um programa que calcula o rendimento diario da minha "caixinha/cofrinho". Pra isso, o codigo precisa lembrar do saldo inserido ontem,
# pedir o saldo de hoje, calcular a diferença e atualizar o valor para o calculo de amanha.

# PASSO 1: IMPORTAR MÓDULOS DO SISTEMA

import os # modulo que conversa com o seu computador pra saber se existe um arquivo salvo no seu PC antes de tentar abri-lo
# vou usar ele como uma trava de segurança para o codigo nao dar erro caso não exista um arquivo.

# PASSO 2: CHECAR E LER O VALOR DO DIA ANTERIOR

# "Python, veja no meu computador se existe o arquivo chamado 'reserva.txt'.
# Se existir, abra ele no modo de leitura ('r'), leia o texto que está lá dentro, 
# converta esse texto para número decimal (float) e guarde na variável 'valor_anterior'.
# Caso o arquivo não exista, defina o valor da variável 'valor_anterior' como None."

if os.path.exists("reserva.txt"):
    with open("reserva.txt", "r") as arquivo:
        valor_anterior = float(arquivo.read())
else:
    valor_anterior = None

# PASSO 3: PEDIR O VALOR DE HOJE
valor = float(input("Quanto ficou sua caixinha hoje? "))

# PASSO 4: CALCULAR O RENDIMENTO E EXIBIR O RESULTADO
if valor_anterior is not None:
    r = valor - valor_anterior
    print(f"Hoje você ganhou R$ {r:.2f}")
else:
    print("Este é o seu primeiro registro! A partir de amanhã farei o cálculo.")

# PASSO 5: ATUALIZAR O ARQUIVO PARA O DIA SEGUINTE
# "Python, abra o arquivo 'reserva.txt' no modo de escrita ('w')
# Em seguida, pegue o valor de hoje, converta ele em texto (str) e escreva dentro do arquivo. 
# Quando terminar,feche o arquivo sozinho (with)."

with open("reserva.txt", "w") as arquivo:
    arquivo.write(str(valor))
```

```python
# O modo 'w' tem um comportamento bem específico: ele sempre limpa o arquivo do zero antes de escrever.
# Ao usar o 'w', o Python joga fora o número velho e grava apenas o número novo de hoje no arquivo.
# Contudo, quero guardar os valores antigos para fazer um gráfico -> usar arquivo JSON
# O JSON permite guardar estruturas de dados do Python (como listas e dicionários) exatamente como elas são
# Dessa forma, o programa nunca apaga nada, ele apenas adiciona (append) um novo registro ao final da lista.
# O json.dump() pega minha lista(historico) da memória RAM e o escreve permanentemente em formato texto (JSON) dentro de um arquivo.
```

```python
import os # conversar com o computador para ver se um arquivo existe
import json # para salvar e ler dados em formato JSON
from datetime import date # para trabahar com data
import plotly # para trabalhar com gráficos

arquivo = 'historico_oficial.json'

if os.path.exists(arquivo):
    with open(arquivo, 'r', encoding='utf-8') as historico_oficial_json:
        historico_oficial = json.load(historico_oficial_json)
else:
    historico_oficial = []

valor_hoje = float(input('Quanto ficou sua caixinha hoje?'))
data_hoje = date.today().strftime('%Y-%m-%d')

if len(historico_oficial) > 0: # se a lista nao for vazia:
    ultimo_registro = historico_oficial[-1] # o ultimo registro vai ser o ultimo item da lista (historico)
    valor_anterior = ultimo_registro['valor'] # o valor_anterior vai ser o valor que esta no ultimo registro(ultimo item da lista)

    r = valor_hoje - valor_anterior
    print(f'Rendimento desde o ultimo registro R$ {r:.2f}')
else:
    r = 0.0
    print('Este é o seu primeiro registro! Histórico iniciado com sucesso!')

novo_registro = {'data': data_hoje, 'valor': valor_hoje, 'rendimento':r}
historico.append(novo_registro)

with open(arquivo, 'w', encoding='utf-8') as f: # abra/crie o arquivo aberto (f) em modo escrita('w') com acentuação correta (utf-8)
    json.dump(historico, f, indent=4) # converte e salva os dados da lista 'historico' dentro do arquivo aberto (f) com 4 linhas de recuo e quebra de linhas para cada dicionario

print('Dados salvos no historico')
```

```python
# Código para acrecentar um dicionário antigo caso vc tenha esquecido de acrescentar (e não queira acrescentar manualmente)

registro_antigo=[
 #  {'data':'2026-07-22','valor':34.55},
]
# adicionar na lista que já está na memoria
historico_oficial.extend(registro_antigo)

#reordenar a lista por data para não ficar bagunçado
historico_oficial.sort(key=lambda item: item['data'])
#sort()= organiza em ordem
#key= como é um dicionario recheado, eu digo por qual campo eu quero ordenar
#lambda item: item['data']: cria uma função rápida de uma linha só, para pegar cada dicionário(item) e ele entra nesse item e pega o valor da chave['data']
# 'Ordene a lista historico usando como critério (key) o valor da chave "data" de cada registro (item)'

#salvar a lista atualizada
with open(arquivo, 'w', encoding='utf-8') as f:
    json.dump(historico_oficial, f, indent=4) 

print('Registro antigo adicionado e arquivo atualizado')
```

```python
# vou acrescentar rendimento nos registros antigos para aparecer no gráfico

arquivo = 'historico_oficial.json'

# 1. Carrega o histórico atual do arquivo
with open(arquivo, 'r', encoding='utf-8') as f:
    historico_oficial = json.load(f)

# 2. Garante que a lista está ordenada por data antes de calcular

historico_oficial.sort(key=lambda item: item['data']) 
# pegue o historico do arquivo, coloque em ordem(.sort) usando como criterio (key) 'data'
# lambda : crie uma função rápida que..
# item: receba um item e me devolve...
# item ['data']: ...o valor contido na chave 'data' desse item


# 3. Percorre a lista e calcula o rendimento de cada registro
for i in range(len(historico_oficial)):
    if i == 0:
        # O primeiro registro do historico não tem anterior, então r é 0.0
        historico_oficial[i]['rendimento'] = 0.0
    else:
        # Para os dias seguintes, pega a diferença do valor atual para o valor do dia anterior
        valor_atual = historico_oficial[i]['valor']
        valor_anterior = historico_oficial[i - 1]['valor']
        
        # Calcula e arredonda para 2 casas decimais
        historico_oficial[i]['rendimento'] = round(valor_atual - valor_anterior, 2)

# len(historico): Descobre quantas linhas(dias) existem na sua lista toda
# range(...): Cria os números das posições para cada dia (ex: 0(dia 1), 1, 2, 3...)
# for i in ...: A variável i assume o valor de cada posição, uma por uma
# i == 0: Para o primeiro dia do histórico, ele define 'rendimento': 0.0
#i > 0: Para cada um dos dias seguintes, ele compara com o registro da posição imediatamente anterior (historico[i - 1])
# round(..., 2): Garante que os rendimentos fiquem bem formatados (ex: 2.21 em vez de 2.210000000002)

# 4. Salva a lista inteira atualizada por cima do JSON
with open(arquivo, 'w', encoding='utf-8') as f:
    json.dump(historico_oficial, f, indent=4) 

print(" Todos os registros antigos foram atualizados com o rendimento!")
```

```python
# criar gráficos, então vou usar o plotly
import json
import plotly.express as px

#carregar os dados do meu arquivo
with open('historico_oficial.json', 'r', encoding='utf-8') as f:
    historico_oficial = json.load(f)

# Preparar dados para o gráfico: preciso organizar os dados no formato que a biblioteca de gráficos (plotly) entende
dados_grafico = [] #Cria uma lista vazia onde vou guardar as informações formatadas para o gráfico

for registro in historico_oficial: #Para cada dicionario (registro) dentro da minha lista historico:
    valor_total = registro['valor'] #Guarda o valor total daquele registro na variável valor_total
    rendimento = registro ['rendimento'] 

# adicionar a parte do valor base:calcular o valor base para ---empilhar---
    dados_grafico.append({ #adiciona um novo dicionario dentro da lista dados_grafico contendo:
        'data': registro['data'], #data do registro
        'tipo_valor':'valor_base', #legenda dizendo o valor investido
        'quantia': valor_total - rendimento # valor base
        })

# adicionar a parte do rendimento: outro dicionario na lista dados_grafico contendo:
    dados_grafico.append({
        'data': registro['data'],
        'tipo_valor': 'rendimento',
        'quantia': rendimento
    })
    
# montar os graficos
grafico = px.bar(
    dados_grafico,
    x='data', 
    color= 'tipo_valor', 
    y='quantia', 
    title='Evolução dos valores por data (Valor base + Rendimento)',
    text_auto=True, # quero que mostre os valores
    barmode='stack' # empilhar as barras
    )

#deixar as barras bem separadinhas e organizadas
grafico.update_xaxes(type='category')
# update_xaxes(): serve para modificar e personalizar a aparência do eixo x do gráfico
# "atualize as propriedades do eixo x com as novas regras que eu mudar(trate as datas como categoria e não calendário)"
grafico.show()

# Salva o gráfico como imagem PNG
grafico.write_image("grafico.png")
 

```
